# What is Generative AI (GenAI)?

## Definition
Generative AI (GenAI) is a type of Artificial Intelligence that can **generate new content** such as text, images, audio, video, or code based on patterns learned from data.

---

## Examples
- Text generation (Chatbots, content writing)
- Image generation (AI art)
- Code generation = CodeLLama
- Music generation
- Video generation like Sora
- TTS models

## 🔍 2) Traditional AI vs Generative AI
- **Traditional AI** focuses on pattern recognition (classification, regression).
- **Generative AI** learns the overall data distribution and produces **new human-like data samples**.

## 3) Application Areas-
- Creative Writing
- Text Summarization
- Software Development
- Customer Support
- Video Generation = Sora

# GenAI vs Agentic AI  
## Backend Engineer Hiring Process (4 Approaches)

---

# 1️⃣ Generative AI (Reactive)

### User says:
> "Hire a backend engineer."

### How it works:

- User: Create JD  
  → GenAI generates JD  

- User: Create LinkedIn post  
  → GenAI generates post  

- User: Screen resume  
  → GenAI summarizes resume  

- User: Generate interview questions  
  → GenAI generates questions  

- User: Draft offer email  
  → GenAI writes offer email  

### Characteristics:
- Fully prompt-driven  
- No external access  
- No memory of long-term goal  
- Human controls every step  

👉 Nature: **Reactive Content Generator**

---

# 2️⃣ RAG-Based Chatbot (Knowledge-Enhanced)

(RAG = Retrieval-Augmented Generation)

### What changes?

- Chatbot connected to:
  - Company hiring policy
  - Past JDs
  - Internal salary bands
  - Tech stack documentation

### How it works:

- User: Hire backend engineer  
  → Retrieves company backend JD template  
  → Generates customized JD  

- Screens resumes  
  → Compares with internal skill matrix  

- Generates interview questions  
  → Based on company tech stack  

### Characteristics:
- Still reactive  
- More accurate & context-aware.(Company Specific = Solved this problem)
- Uses internal knowledge base  
- Cannot execute real-world actions  

👉 Nature: **Smarter Assistant (Knowledge + LLM)**

---

# 3️⃣ Tool-Augmented Chatbot (Action-Enabled)

### What changes?

- Connected to tools/APIs:
  - LinkedIn API  
  - Job portals  
  - Email API 
  - Calendar API
  - Human Research Tool

### How it works:

- User: Post job  
  → Generates JD  
  → Posts via LinkedIn API  

- User: Schedule interview  
  → Books calendar slot automatically  

- User: Send offer  
  → Sends email via Email API  

### Characteristics:
- Can execute actions (Solved)
- Still requires human prompts **REACTIVE** (Problem)
- No long-term autonomous planning  
- Acts only when instructed  

👉 Nature: **Action-Capable Assistant**

---

# 4️⃣ Fully Agentic AI (Autonomous System)

### User says:
> "Hire a backend engineer."

### What it does:

1. Understands goal  
2. Creates hiring plan  
3. Generates JD  
4. Posts job automatically  
5. Screens resumes  
6. Shortlists candidates  
7. Schedules interviews  
8. Evaluates results  
9. Adapts strategy if needed  
10. Sends final offer (asks for approval)

### Characteristics:
- Proactive = Goal Identify and Plan
- Adaptability
- Memory
- Context Aware
- Tool Support

👉 Nature: **Autonomous Goal Executor**

---

# 🔥 Clear Progression

| Level | Capability |
|--------|------------|
| GenAI | Generates content |
| RAG-Based | Generates with company knowledge |
| Tool-Augmented | Executes actions when told |
| Agentic AI | Plans + Acts + Adapts autonomously |

---
# GenAI vs Agentic AI

| Aspect | Generative AI (GenAI) | Agentic AI |
|----------|------------------------|-------------|
| Main Focus | Content creation | Goal completion |
| Purpose | Creates text, images, code | Solves tasks end-to-end |
| Behavior | Reactive | Proactive (autonomous) |
| Action Taking | No real-world actions | Can plan & execute actions |
| Relationship | Building block | Uses GenAI + memory + tools + planning |
